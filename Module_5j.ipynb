## Tuples

```python
# declaration

tup = (10,20, 1, 3, 4)

float_tup = (10.3, 33, 40, 2)

print(type(tup), tup)

print(type(float_tup), float_tup)

mix_tup = (10, 20.22, "str", True)

print(type(mix_tup), mix_tup)
```

```python
lst =tuple([10, 20, 30, 49]) ## list to tuple

tup1 = list[(10.2, 30, 20, 11)]

print(type(tup1), tup1)

print(type(lst), lst)
tup = (10,)

print(type(tup), tup)
```

```python
## access

# index
tup =(10, 30, 10.20, "str", True)

print(tup[3])

# access by slicing
new_tup = tup[0:3]

print(new_tup)
print(new_tup[0:1])

```

```python
# muteability

lst = [1, 20, 30 , 44, 101]
lst.append(55)
lst[1] = 200
print(lst)


# immutable 

tup = (10, 20, 10.1, 40)
#tup[1] = 100 # is not possible to to add anything

print(tup)
```

```python
# tuple method

tup = (10, 20 ,30 , 10 , 30, 50, 29, 88)

print(tup.count(10))
print(tup.index(50))
```

## Set Mathematical Method

```python
a = {1, 2, 3}
b = {2, 3, 4, 5}

print(a.union(b))

print(a.intersection(b))

x = {1, 2, 3,}
y = {1, 2, 3, 4, 5}
print(x.isdisjoint(y))

print(x.issubset(y))
```

## Dictionary

```python
# Dictionary = mutable , unorderd ( so we cannot use indexing)
#Store data as a (key, value)
# key = immutalbe and we can keep (int, string, float, tuple)
# value , we can keep (string, lsit , set, string, int, dictionary

# declaration
dic = {}
print(type(dic),dic)

# value assin
# key : value
dic = {'name' : 'Phitron', 'age': 28, 'address':'Dhaka', 'number': [17, 23, 2, 3]}

print(type(dic), dic)
```

```python
# access
# dic_name[key]

print(dic["address"])
print(dic.get("number")) #another way to get access

# change the value

dic["name"] = "Phitron AI/ML"

# doulbe key = also take latest key if input double
dic = {'name' : 'Phitron', 'age': 28, 'address':'Dhaka', 'number': [17, 23, 2, 3], 'age': 20}

print(dic)
```

```python
# printing default value

print(dic)

print(dic.get("math_makrs"))
print(dic.get("math_marks",88))
print(dic)
```

```python
# adding a value
dic = {'name' : 'Phitron', 'age': 28, 'address':'Dhaka', 'number': [17, 23, 2, 3], 'age': 20}

dic["math_marks"] = 88
print(dic)

dic.update({"english_marks": 92})
print(dic)
```

```python
# deleting
del dic["math_marks"]
print(dic)
```

```python
# copy
dic_2 = dic.copy()
print(dic_2)
```

## Iterating over Dictionary

```python
dic = {'name' : 'Phitron', 'age': 28, 'address':'Dhaka', 'number': [17, 23, 2, 3], 'age': 20}

keys = dic.keys()
print(keys)

dic ["math_marks"] = 80
print(keys)

values = dic.values()
print(values)

dic ["math_marks"] = 70
print(values)


items = dic.items()
print(items)
```

```python
# iteration
for key,value in dic.items():

    print(key,value)
```

## Dictionary Comprehension 

```python
square = {x:x**2 for x in range(1,11) if x%2==0}
print(square)
```

```python
co_ordinate = [(10, 10.5), (28, 192), (108,99)]
location = ["dhaka", "chattrogram","sylhet"]
exect_location = {co_or : loc for co_or,loc in zip(co_ordinate,location)}

print(exect_location)
```

```python
subject =["bangla", "english", "math", "accounting", "physics"]
teacher = ["saim", "pankaj", "dulal", "belly", "anyone"]

class_routine = {sub : tea for sub,tea in zip(subject,teacher)}
print(class_routine)
```
