## String Method

```python
string ="Welcome to Phitron ML Course in, phitron ML to ML to ML"

#process_strings = string.upper()
#process_strings = string.capitalize()
process_strings = string.lower()

print(process_strings)
#Length
print(len(string))

# word ase kina -> bool
print("Phitron" in string)

# Shurur kon index theke substring ta shuru hoise

shuru_index = process_strings.find("phitron")
print(shuru_index)

#Sesher kon index theke substring ta shuru hoise

sesher_index = process_strings.rfind("phitron")
print(sesher_index)


# Substring koi bar ache
count = process_strings.count("to")
print(count)

# replace a substring

process_string = string.replace("ML", "AI/ML")
print(process_string)
```
