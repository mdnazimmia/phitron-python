```python
def square_addition(a,b):
    a=a**2
    b=b**2
    print(a+b)
square_addition(2,4) # if i want to do more mathmatic is not possible
```

```python
def square_add(a,b,c):
    print(f"{a},{b},{c}")
    a=a**2
    b=b**2
    return(a+b+c)

# position argument
add = square_add(10,20,30) + 10
print(add)

# keyword argument
add = square_add(b=10,c=20,a=30) + 10
print(add)
```

```python
# position arguments
def square_add(*args):

    summation = 0
    
    for i in args:
        i=i**2
        summation+=i
    return summation
    
ans= square_add(10,20,30,40,50)
print(ans)
```

```python
# Key arguments
def student(**kyarg):
    for key,val in kyarg.items():
        print(f"{key} : {val}")
```

```python
student(naem ='adil', roll = 5, clas = 8, age =13, marks = 78)
```
