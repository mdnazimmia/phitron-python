```python
stack = []

# Stack a push
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

# Getting the top element
#print(stack.pop())
#print(stack.pop())
#print(stack.pop())
#print(stack.pop())
#print(stack.pop())

print(stack.pop())

# access the first element
print(f"In front {stack[-1]}\n")

print(stack.pop())
print(f"In front {stack[-1]}\n")
```

## List comprehension

```python
even = []

for i in range(101):
    if i % 2 == 0:
        even.append(i)

print(even)
```

```python
random_list =[x for x in range(1,101) if x % 2 == 0]

print(random_list)
```

```python
fruits = ["Mango", "Lichi", "Banana", "Guaba"]

upper_fruits=[]

for fruit in fruits:
    upper_fruits.append(fruit.upper())

print(upper_fruits)
```

```python
fruits = ["Mango", "Lichi", "Banana", "Guaba"]

lower_fruits =[fruit.lower() for fruit in fruits]

print(lower_fruits)
```
