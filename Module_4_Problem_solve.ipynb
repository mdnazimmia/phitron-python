## Smart Llight Pilot

```python
inp = input()
numbers = inp.split()

x = float(numbers[0])
y = float(numbers[1])
if x >= y:
    print("ON")
else:
    print("OFF")
```

##  Sensor Calibrator — Min-Max Scaling

```python
inp = input()
calibrating = inp.split()

x = float(calibrating[0])
min_v = float(calibrating[1])
max_v = float(calibrating[2])

norm = (x - min_v) / (max_v - min_v)
print(f"{norm:.2f}")
```
