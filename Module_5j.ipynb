## Tuples

```python
# declaration

tup = (10,20, 1, 3, 4)

float_tup = (10.3, 33, 40, 2)

print(type(tup), tup)

print(type(float_tup), float_tup)

mix_tup = (10, 20.22, "str", True)

print(type(mix_tup), mix_tup)
```

```python
lst =tuple([10, 20, 30, 49]) ## list to tuple

tup1 = list[(10.2, 30, 20, 11)]

print(type(tup1), tup1)

print(type(lst), lst)
tup = (10,)

print(type(tup), tup)
```

```python
## access

# index
tup =(10, 30, 10.20, "str", True)

print(tup[3])

# access by slicing
new_tup = tup[0:3]

print(new_tup)
print(new_tup[0:1])

```

```python
# muteability

lst = [1, 20, 30 , 44, 101]
lst.append(55)
lst[1] = 200
print(lst)


# immutable 

tup = (10, 20, 10.1, 40)
#tup[1] = 100 # is not possible to to add anything

print(tup)
```

```python
# tuple method

tup = (10, 20 ,30 , 10 , 30, 50, 29, 88)

print(tup.count(10))
print(tup.index(50))
```

## Set

```python
## unique elements (note : so not possible to keep duplicate value)
## mutable ( note: not possible to change value)
## unorderd ( note: indexing & slicing not possible)

A = {1, 2, 3, 2, 1, 4, 5}
print(type(A),A)
```

```python
lst = []

print(type (lst), lst)

tup = ()
print(type(tup), tup)

s = {}
print(type(s),s)
```

```python
S = {1, 2, 3, 4, 5, 1 , 3 , 2}

if (10 in S):
    print("3 is present")
else:
    print(" is not present")


for element in S:
    print(element)

sum = 0

for element in S:
    sum += element
print(sum)

    
```

```python
# mutable 
s = {10, 1, 2, 9, 3, 2, 4, 6, 8}
print(s)

s.add(11)
s.add(0)
print(s)

s.pop()
s.pop()

print(s)

s.remove(8)
print(s)


```

## Set Mathematical Method

```python
a = {1, 2, 3}
b = {2, 3, 4, 5}

print(a.union(b))

print(a.intersection(b))

x = {1, 2, 3,}
y = {1, 2, 3, 4, 5}
print(x.isdisjoint(y))

print(x.issubset(y))
```
