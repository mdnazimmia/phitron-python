## Function

```python
# Position Parameters
def greeting():
    print("Welcome to conceptual Session 1")
greeting()
```

```python
def greeting(x,y):
    print("Welcome to conceptual session - 1.", x, y)
greeting("Noor", "Abdul")
```

```python
# Keyword Parameter
# country default Parameter
def info ( name, age, country = "Bangladesh"):
    print(f"{name} age is {age} and country {country}")

info(age = 30, name = "Nazim")
```

```python
# Variable length argument
def add (a, b, c):
    return a + b + c
print(add(1, 2, 3))
```

```python
# Variable length argument
def add (*args):
    print(type(args))
    print(args)
add(1,2,3,4,5)
```

```python
# Keyword
def add (**k):

    print(f"{k['name']} age is : {k['age']} and country is {k['country']}")
add(name = "Nazim", age =30, country = "Bangladesh")
```

```python
# hints
def func() -> int:
    return 10
print(func())
```

```python
# Dictionary Return
def get_student():
    student = {
        "name": "Nazim",
        "age": 30,
        "city": "Brahmanbaria"
    }
    return student
info = get_student()
print(info)
```
